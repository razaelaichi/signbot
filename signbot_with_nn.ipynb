```python
import pandas as pd
import torch
```

```python
file_path = 'data.csv'
```

```python
data = pd.read_csv(file_path)
```

```python
x = data['Input (x)'] ; y = data['Label (y)']
x
```

```python
g = torch.Generator().manual_seed(123123123)
W = torch.randn(generator = g, size = (1,1), requires_grad=True)
b = torch.tensor(0.01, requires_grad=True)
```

```python
b
```

```python
#input tensor
inp_tns = torch.tensor(x.values, dtype= torch.float32, requires_grad=True).view(-1, 1)
inp_tns
```

```python
#label tensor
lbl_tns = torch.tensor(y.values).view(-1,1)
lbl_tns.shape
```

```python
logits = (inp_tns @ W) + b
```

```python
predictions = torch.sigmoid(logits)
```

```python
diff = lbl_tns - predictions
sq_diff = diff**2
```

```python
loss_avg = sq_diff.mean()
loss_avg.backward()
```

```python
W.grad
```

```python
b.grad
```

```python
for i in range(100000):
    lgt_neu = (inp_tns @ W) + b
    prdc_neu = torch.sigmoid(lgt_neu)
    l_avg_neu = ((lbl_tns - prdc_neu)**2).mean()
    l_avg_neu.backward()
    with torch.no_grad():
        W -= 0.1 * W.grad
        b -= 0.1 * b.grad
        print(l_avg_neu)
        W.grad.zero_()
        b.grad.zero_()
```

```python
l_avg_neu.item()
```

```python
## giving new inputs to get answers from the trained neuron
x_dash = torch.tensor(float(input('Enter your number : ')), dtype=torch.float32)
accuracy = torch.sigmoid((W * x_dash) + b).item()
if accuracy > 0.5:
    print(f'Positive number, predicted with {accuracy*100}%')
else:
    print(f'Negative number, predicted with {(1-accuracy)*100}%')
```
